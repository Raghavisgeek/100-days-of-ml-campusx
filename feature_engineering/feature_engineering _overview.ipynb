# 1.4 feature scaling
    feature scaling: According to the scikit-learn Preprocessing Guide, unscaled data can cause variables with large numbers to dominate models that rely on distances or gradients:
1. Prevents Bias: Keeps a large feature (like income in millions) from hiding a small feature (like age in single digits).
2. Speeds Up Training: Helps optimization tools like gradient descent reach the best answer much faster.
3. Improves Accuracy: Essential for distance-based models like K-Nearest Neighbors (KNN), Support Vector Machines (SVM), and Neural Networks.
    
feature scaling prevents the domination of one feature over other so that model doesnt become bias
